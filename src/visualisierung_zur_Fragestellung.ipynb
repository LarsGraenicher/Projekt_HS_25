```python
data="mydata"

fragestellung = alt.Chart(data,width= 800, height = 500).transform_calculate(
    PercentOfadult_ltr_pedestrians="datum.adult_ltr_pedestrians_count/datum.adult_pedestrians_count"
).transform_aggregate(
    mean_percent="mean(PercentOfadult_ltr_pedestrians)",
    sum_adults="sum(adult_pedestrians_count)",
    sum_ltr="sum(adult_ltr_pedestrians_count)",
    groupby=["location_name"]
).transform_window(
    rank="rank(mean_percent)",
    sort=[{"field": "mean_percent", "order": "descending"}]
).mark_bar(size=140,).encode(
    alt.Y('mean_percent:Q', ).axis(title="Anteil der Erwachsenen die zum Bahnof gehen", titleFontSize=18, format='.0%', labelFontSize=16, tickCount=7, ),  
    #tickCount sorgt dafür das es höchstens 7ben Klassen gibt, bei AltScale gibt das Padding an das die Balken nicht bis zu oberst gehen sollen.
    alt.X('location_name:N').axis(title="Untersuchte Strassen",  titleFontSize=18, labelAngle=0, labelFontSize=16), color=alt.condition(
    "datum.rank == 1",
    alt.value("red"),
    alt.value("lightgray")
), tooltip=[
        alt.Tooltip('sum_adults:Q', title='Anzahl Erwachsene insgesamt'),#alt.Tooltip wird dafür benötigt um einen Titel und so zu vergeben.
        alt.Tooltip('sum_ltr:Q', title='Anzahl Erwachsene nach ltr'),
    
    ],
    )
fragestellung
```

```python
import json

fragestellung_json= fragestellung.to_dict()

with open("jsons/fragestellung_spec.json", "w") as f: #zum speichern
    json.dump(fragestellung_json, f, indent=2)
```
